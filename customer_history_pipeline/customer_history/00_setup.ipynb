```python
spark.sql("CREATE SCHEMA IF NOT EXISTS workspace.bank_project")

spark.sql("""
          
CREATE TABLE IF NOT EXISTS workspace.bank_project.customer_changes (
    customer_id     STRING,
    full_name       STRING,
    city            STRING,
    risk_rating     STRING,
    kyc_status      STRING,
    operation       STRING,
    change_time     TIMESTAMP,
    batch_id        INT
)
""")


spark.sql("""
CREATE TABLE IF NOT EXISTS workspace.bank_project.dim_customer_manual (
    customer_id     STRING,
    full_name       STRING,
    city            STRING,
    risk_rating     STRING,
    kyc_status      STRING,
    valid_from      TIMESTAMP,
    valid_to        TIMESTAMP,
    is_current      BOOLEAN
)
""")


print("Setup_Done!")

```
